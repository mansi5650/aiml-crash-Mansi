```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def loss(theta):
    return (theta - 3.5)**2 + 1

```

```python
def gradient(theta):
    return 2 * (theta - 3.5)
```

```python
theta = 8
learning_rate = 0.1

results = []

for i in range(15):
    g = gradient(theta)
    l = loss(theta)

    results.append([i, theta, l, g])

    theta = theta - learning_rate * g

df = pd.DataFrame(
    results,
    columns=["Iteration", "Theta", "Loss", "Gradient"]
)

df
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Iteration"], df["Loss"], marker="o")
plt.title("Loss Curve (Learning Rate = 0.1)")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```

```python
theta = 8
learning_rate = 0.01

loss_low = []

for i in range(15):
    loss_low.append(loss(theta))
    theta = theta - learning_rate * gradient(theta)
```

```python
theta = 8
learning_rate = 0.5

loss_high = []

for i in range(15):
    loss_high.append(loss(theta))
    theta = theta - learning_rate * gradient(theta)
```

```python
theta = 8
learning_rate = 0.1

loss_medium = []

for i in range(15):
    loss_medium.append(loss(theta))
    theta = theta - learning_rate * gradient(theta)

plt.figure(figsize=(8,5))

plt.plot(loss_low, marker="o", label="LR = 0.01")
plt.plot(loss_medium, marker="s", label="LR = 0.1")
plt.plot(loss_high, marker="^", label="LR = 0.5")

plt.title("Comparison of Learning Rates")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

## Comparison of Learning Rates

1. A low learning rate (0.01) decreases the loss slowly.

2. A medium learning rate (0.1) converges faster and smoothly.

3. A higher learning rate (0.5) reaches the minimum much faster for this simple function.

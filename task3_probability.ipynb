```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
uniform_data = np.random.uniform(0, 10, 1000)

print("First 10 Uniform Values:")
print(uniform_data[:10])

```

```python
normal_data = np.random.normal(5, 2, 1000)

print("First 10 Normal Values:")
print(normal_data[:10])
```

```python
print("----- Uniform Distribution -----")
print("Mean:", np.mean(uniform_data))
print("Standard Deviation:", np.std(uniform_data))

print()

print("----- Normal Distribution -----")
print("Mean:", np.mean(normal_data))
print("Standard Deviation:", np.std(normal_data))
```

```python
plt.figure(figsize=(8,5))
plt.hist(uniform_data, bins=30)
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(normal_data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

## Statistics Summary and Interpretation

### Uniform Distribution
- Mean = approximately 5
- Standard Deviation = approximately 2.9

### Normal Distribution
- Mean = approximately 5
- Standard Deviation = approximately 2

### Interpretation

1. The uniform distribution spreads values almost equally across the range.

2. The normal distribution has most values concentrated around the mean, creating a bell-shaped curve.

3. The normal distribution is commonly used in Machine Learning and statistics because many real-world datasets approximately follow this pattern.

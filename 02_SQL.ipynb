```python
import sqlite3
import pandas as pd
```

```python
df = pd.read_csv('final_merged_df.csv')
```

```python
df.head()
```

```python
conn = sqlite3.connect('ecommerce.db')
```

```python
df.to_sql(
    'orders',
    conn,
    if_exists='replace',
    index=False
)
```

```python
query = """
SELECT COUNT(*) AS total_orders
FROM orders;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT *
FROM orders
LIMIT 5;
"""

pd.read_sql(query, conn)
```

## Query 1: Total Number of Orders

Purpose: To find the total number of orders in the dataset.

```python
query = """
SELECT COUNT(*) AS total_orders
FROM orders;
"""

pd.read_sql(query, conn)
```

## Query 2: Delayed Orders

Purpose: To determine how many orders were delivered late.

```python
query = """
SELECT COUNT(*) AS delayed_orders
FROM orders
WHERE delay_days > 0;
"""

pd.read_sql(query, conn)
```

## Query 3: Average Delay Days

Purpose: To calculate the average delay in delivery.


```python
query = """
SELECT AVG(delay_days) AS average_delay
FROM orders;
"""

pd.read_sql(query, conn)
```

## Query 4: Payment Method Distribution

**Purpose:** To determine the number of orders made using each payment method.


```python
query = """
SELECT payment_type,
       COUNT(*) AS total_orders
FROM orders
GROUP BY payment_type;
"""

pd.read_sql(query, conn)
```

## Query 5: Top 10 Customer States

**Purpose:** To identify the states with the highest number of orders.

```python
query = """
SELECT customer_state,
       COUNT(*) AS total_orders
FROM orders
GROUP BY customer_state
ORDER BY total_orders DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

## Query 6: Average Delivery Days

**Purpose:** To calculate the average number of days taken for delivery.

```python
query = """
SELECT AVG(delivery_days) AS avg_delivery_days
FROM orders;
"""

pd.read_sql(query, conn)
```

## Query 7: Monthly Order Trend

**Purpose:** To analyze the number of orders placed each month.

```python
query = """
SELECT month,
       COUNT(*) AS total_orders
FROM orders
GROUP BY month
ORDER BY month;
"""

pd.read_sql(query, conn)
```

## Query 8: States with More Than 1000 Orders

**Purpose:** To find states with a high concentration of orders.


```python
query = """
SELECT customer_state,
       COUNT(*) AS total_orders
FROM orders
GROUP BY customer_state
HAVING COUNT(*) > 1000;
"""

pd.read_sql(query, conn)
```

## Query 9: Maximum Delay

**Purpose:** To identify the highest delivery delay recorded.

```python
query = """
SELECT MAX(delay_days) AS max_delay
FROM orders;
"""

pd.read_sql(query, conn)
```

## Query 10: Top 5 Payment Methods

**Purpose:** To determine the five most frequently used payment methods.



```python
query = """
SELECT payment_type,
       COUNT(*) AS total_orders
FROM orders
GROUP BY payment_type
ORDER BY total_orders DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Function
def f(x):
    return x**2 + 2*x + 1
    
```

```python
# Numerical Derivative
def numerical_derivative(x):
    h = 0.0001
    return (f(x + h) - f(x)) / h
```

```python
# Calculate derivative at three points
points = [-2, 0, 2]

print("Numerical Derivatives:\n")

for p in points:
    print(f"x = {p}, Derivative = {numerical_derivative(p):.4f}")
```

```python
# Generate x and y values
x = np.linspace(-5, 5, 400)
y = f(x)
```

```python
# Plot function and tangent lines

plt.figure(figsize=(8,6))
plt.plot(x, y, label="f(x)=x²+2x+1", linewidth=2)

for p in points:
    slope = numerical_derivative(p)
    y_point = f(p)

    tangent = slope * (x - p) + y_point

    plt.plot(x, tangent, "--", label=f"Tangent at x={p}")
    plt.scatter(p, y_point)

plt.title("Function and Tangent Lines")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)

plt.show()
```

### Observations

1. The slope increases as the value of x increases.

2. A negative slope means the function is decreasing, while a positive slope means it is increasing.

3. In Machine Learning optimization, a near-zero slope indicates that the model is close to the minimum loss.
